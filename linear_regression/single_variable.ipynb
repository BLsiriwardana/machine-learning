```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
data = pd.read_csv("Book.csv") #import data set from csv file via using "pandas"
```

```python
data
```

```python
plt.scatter(data.videos,data.views) #give data set for x & y axis 
plt.xlabel('Number of videos') # named axis 
plt.ylabel('Total Views')
```

```python
print(type(data.videos))
x = np.array(data.videos.values)
x
```

```python
y = np.array(data.views.values)
y
```

```python
model = LinearRegression()
model.fit(x.reshape((-1,1)),y)
```

```python
new_x = np.array([45]).reshape((-1,1))
new_x
```

```python
predict = model.predict(new_x)
predict
```
